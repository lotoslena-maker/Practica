````python
#2. Побудова моделі
from sklearn.linear_model import LinearRegression

# Ознака та цільова змінна
features = ['day']
target = 'tank'

# Навчальні дані
X_train = train_set[features]
y_train = train_set[target]

# Тестові дані
X_test = test_set[features]
y_test = test_set[target]

# Створення і навчання моделі
model = LinearRegression()
model.fit(X_train, y_train)

# Прогноз
predictions = model.predict(X_test)

print("Перші 10 прогнозів:")
print(predictions[:10])
````

````python
import pandas as pd

df = pd.read_csv(
    'archive/russia_losses_equipment.csv'
)

display(df.head())
print(df.columns.tolist())
````

````python
# Потім побудуйте модель на початкових, ненормалізованих значеннях:

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Вибираємо потрібні стовпці та видаляємо пропуски
model_data = df[['day', 'tank']].dropna().sort_values('day')

# Хронологічний поділ 80/20
split_index = int(len(model_data) * 0.8)

train_data = model_data.iloc[:split_index]
test_data = model_data.iloc[split_index:]

X_train = train_data[['day']]
y_train = train_data['tank']

X_test = test_data[['day']]
y_test = test_data['tank']

# Навчання моделі
model = LinearRegression()
model.fit(X_train, y_train)

# Прогнозування
predictions = model.predict(X_test)

# Оцінювання
mae = mean_absolute_error(y_test, predictions)
mse = mean_squared_error(y_test, predictions)
r2 = r2_score(y_test, predictions)

print(f'MAE: {mae:.2f}')
print(f'MSE: {mse:.2f}')
print(f'R²: {r2:.3f}')
````

````python
# Візуалізація:

results = pd.DataFrame({
    'day': test_data['day'],
    'Фактичне значення': y_test,
    'Прогноз моделі': predictions
})

import plotly.express as px

fig = px.line(
    results,
    x='day',
    y=['Фактичне значення', 'Прогноз моделі'],
    title='Фактичні та прогнозовані накопичені втрати танків'
)

fig.show()
````

Модель правильно сформувала прогноз, проте його точність є недостатньою. Прогнозовані значення систематично перевищують фактичні, а розбіжність зростає з часом. Це свідчить про те, що проста лінійна залежність від номера дня не враховує зміну темпу накопичення показника. Для підвищення точності доцільно навчати модель на актуальнішому часовому проміжку або використовувати додаткові ознаки.

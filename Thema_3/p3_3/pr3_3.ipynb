```python

import pandas as pd

   # Завантаження даних
df = pd.read_csv('operations_db.csv')

   # Перегляд перших рядків
print(df.head())
   
   # Огляд інформації про набір даних
print(df.info())
```

```python
# 1. Перевірка пропущених значень
print("Кількість пропущених значень у кожному стовпці:")
print(df.isnull().sum())
```

```python
# 2. Заповнення пропущених значень

# Числові стовпці заповнюємо медіаною
numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
#3. Видалення дублікатів
df.drop_duplicates(inplace=True)
```

```python
 #4 Перетворення категоріальних змінних на числові (якщо потрібно)
df['category_column'] = df['category_column'].astype('category').cat.codes
```

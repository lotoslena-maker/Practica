````python
# Описова статистика
print(df.describe())
````

count — кількість значень;
mean — середнє арифметичне;
std — стандартне відхилення;
min — мінімальне значення;
25% — перший квартиль;
50% — медіана;
75% — третій квартиль;
max — максимальне значення.

````python
# Вибираємо лише числові стовпці
numeric_df = df.select_dtypes(include='number')

# Основні статистичні показники
statistics = pd.DataFrame({
    'Середнє значення': numeric_df.mean(),
    'Медіана': numeric_df.median(),
    'Стандартне відхилення': numeric_df.std()
})

display(statistics)
````

````python
# без стопчика, що перетворили на числовий
numeric_df = df.select_dtypes(include='number').drop(
    columns=['location_1_code'],
    errors='ignore'
)

statistics = pd.DataFrame({
    'Середнє значення': numeric_df.mean(),
    'Медіана': numeric_df.median(),
    'Стандартне відхилення': numeric_df.std()
})

display(statistics)
````

````python
correlation_matrix = df[['start_date', 'end_date']].corr()

display(correlation_matrix)
````

````python
import plotly.express as px

fig = px.imshow(
    correlation_matrix,
    text_auto='.2f',
    color_continuous_scale='RdBu_r',
    zmin=-1,
    zmax=1,
    title='Кореляційна матриця'
)

fig.show()
````

````python
r = correlation_matrix.loc['start_date', 'end_date']

print(f'Коефіцієнт кореляції: {r:.2f}')
````

Орієнтовне тлумачення:

близько до 1 — сильний прямий зв’язок;
близько до −1 — сильний обернений зв’язок;
близько до 0 — лінійний зв’язок майже відсутній.



Проведено кореляційний аналіз числових змінних start_date та end_date. Отриманий коефіцієнт кореляції характеризує напрям і силу лінійного зв’язку між датами початку та завершення операцій. Водночас кореляція не доводить причинно-наслідкового зв’язку.

````python
# Візуалізація кореляційної матриці по іншому
import plotly.express as px

# Вибір числових стовпців
numeric_df = df.select_dtypes(include='number')

# Розрахунок кореляцій
correlation_matrix = numeric_df.corr()

# Виведення таблиці
display(correlation_matrix)

# Візуалізація
fig = px.imshow(
    correlation_matrix,
    text_auto='.2f',
    color_continuous_scale='RdBu_r',
    zmin=-1,
    zmax=1,
    title='Кореляційна матриця'
)

fig.show()
````
